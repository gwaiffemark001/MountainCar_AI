# MountainCar AI: Reinforcement Learning Report

This notebook documents the completed MountainCar-v0 project. The implementation covers value iteration, MDP transformation, model-based Monte Carlo, tabular Q-learning, Fourier features, function approximation, and constrained exploration.

**Team scope:** Person 1 implemented the AI algorithm and environment integration. AI assistance was used for implementation support and testing, and the final code was reviewed and executed locally.


## Environment

- State: continuous `(position, velocity)`.
- Actions: push left (`0`), no push (`1`), push right (`2`).
- Reward: `-1` per step until the car reaches position `0.5`.
- Horizon: 200 steps.
- Tabular representation: 40 position bins x 40 velocity bins x 3 actions.


## Core methods

**Value iteration** repeatedly applies the Bellman optimality update:
`V(s) = max_a sum_{s'} P(s'|s,a)[R(s,a,s') + gamma V(s')]`.

**MDP transformation** removes discounting by multiplying continuation probabilities by `gamma`, scaling continuation rewards by `1/gamma`, and adding an absorbing `OUT` state with probability `1-gamma`.

**Q-learning** updates one state-action estimate from observed transitions. Function approximation replaces the table with linear weights over Fourier features. Constrained Q-learning filters actions whose predicted velocity would exceed the safety threshold.
